```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Remove duplicate rows if any
df = df.drop_duplicates()

# Fill missing numerical values with median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Check missing values again
print("\nMissing values after cleaning:")
print(df.isnull().sum().sum())

print("\nDataset shape after cleaning:", df.shape)
```

```python
print("Number of farms:", df['Farm_ID'].nunique())
print("Number of states:", df['State'].nunique())
print("Number of districts:", df['District'].nunique())
print("Number of crops:", df['Crop'].nunique())
print("Seasons:", df['Season'].unique())

print("\nCrop distribution:")
print(df['Crop'].value_counts())

print("\nSeason distribution:")
print(df['Season'].value_counts())
```

```python
df.describe()
```

```python
season_summary = df.groupby('Season').agg({
    'Yield_Tonnes_Ha': 'mean',
    'Production_Tonnes': 'mean',
    'Revenue_INR': 'mean',
    'Profit_INR': 'mean',
    'Water_Used_m3': 'mean',
    'Water_Efficiency_t_per_1000m3': 'mean',
    'Disease_Pest_Risk_pct': 'mean'
}).round(2)

season_summary
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='Season',
    y='Yield_Tonnes_Ha',
    estimator='mean'
)

plt.title('Average Crop Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.show()
```

```python
crop_season_yield = df.pivot_table(
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

crop_season_yield
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt='.2f'
)

plt.title('Average Crop Yield Across Seasons')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.show()
```

```python
corr_cols = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_Moisture_pct',
    'Nitrogen_kg_ha',
    'Phosphorus_kg_ha',
    'Potassium_kg_ha',
    'Fertilizer_kg_ha',
    'Seed_Quality_Score',
    'Yield_Tonnes_Ha',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

plt.figure(figsize=(12,8))

sns.heatmap(
    df[corr_cols].corr(),
    annot=True,
    fmt='.2f'
)

plt.title('Correlation Between Agricultural Factors and Performance')
plt.show()
```
